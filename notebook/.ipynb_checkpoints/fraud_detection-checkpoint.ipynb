## Section 2 — Load Dataset

# Dataset Overview
| Property | Value |
|---|---|
| Source | ULB Credit Card Fraud Detection (Kaggle, 2018) |
| Raw Rows | 284,807 transactions |
| Features | 30 (Time, V1–V28, Amount) + 1 target (Class) |
| Fraud Cases | 492 (0.17%) |
| Class | 0 = Legitimate, 1 = Fraudulent |

# Data Notes
- Features **V1–V28** are the result of a PCA transformation applied by the dataset authors to protect cardholder privacy. The original feature names are not available.
- **Time** represents seconds elapsed since the first transaction in the dataset.
- **Amount** represents the transaction value in Euros.

# Duplicate Removal
The dataset contains duplicate rows which must be removed prior to any analysis to prevent data leakage and inflated evaluation metrics.

```python
# =============================================================================
# SECTION 2 — LOAD DATASET
# =============================================================================

# Load the raw dataset from the local data directory
df = pd.read_csv('../data/creditcard.csv')

# Display the shape of the raw dataset
print(f"Raw dataset shape: {df.shape}")

# Display the first five rows to confirm successful loading
print("\nFirst five rows:")
df.head()
```

```python
# Check for duplicate transactions
duplicates = df.duplicated().sum()
print(f"Duplicate rows found: {duplicates}")

# Remove duplicate rows and reset the index
df = df.drop_duplicates()
df = df.reset_index(drop=True)

# Confirm the cleaned dataset shape
print(f"Dataset shape after removing duplicates: {df.shape}")
```

Exploratory Data Analysis is conducted to understand the structure, distribution, and characteristics of the dataset prior to preprocessing. This analysis directly informs and justifies the preprocessing decisions documented in AT2.

# Key Objectives

## Section 4 — Preprocessing

# Purpose
Raw data must be preprocessed before being passed to machine learning models. This section applies the preprocessing pipeline defined in AT2.

# Preprocessing Steps
| Step | Detail |
|---|---|
| Feature/target separation | Class column separated from feature matrix |
| Feature scaling | StandardScaler applied to Amount and Time only |
| Train/test split | Stratified 80/20 split with random seed 42 |

# Scaling Decision
**V1–V28 are not scaled.** These features were already standardised during the PCA transformation applied by the dataset authors. Scaling them again would distort the existing transformation and introduce noise.

**Amount and Time are scaled** using StandardScaler, which transforms each feature to have a mean of 0 and a standard deviation of 1. This prevents large-magnitude features from dominating distance-based calculations within the models.

# Stratified Split
Stratification ensures that the 0.17% fraud ratio is preserved proportionally in both the training and test sets. Without stratification, random splitting risks placing too few or zero fraud cases in one of the sets.

```python
# =============================================================================
# SECTION 4 — PREPROCESSING
# =============================================================================

# Separate features and target variable
X = df.drop('Class', axis=1)
y = df['Class']

print(f"Features shape: {X.shape}")
print(f"Target shape:   {y.shape}")
```

```python
# Apply StandardScaler to Amount and Time only
# V1-V28 are already PCA-transformed and do not require further scaling
scaler = StandardScaler()
X[['Amount', 'Time']] = scaler.fit_transform(X[['Amount', 'Time']])

print("StandardScaler applied to Amount and Time.")
print("\nScaled statistics for Amount and Time:")
X[['Amount', 'Time']].describe().round(4)
```

```python
# Perform stratified 80/20 train-test split
# Stratified ensures the fraud ratio is preserved in both sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=RANDOM_SEED,
    stratify=y
)

print(f"Training set shape:   {X_train.shape}")
print(f"Test set shape:       {X_test.shape}")
print(f"\nTraining set class distribution:")
print(f"  Legitimate: {(y_train == 0).sum():,}")
print(f"  Fraud:      {(y_train == 1).sum():,}")
print(f"\nTest set class distribution:")
print(f"  Legitimate: {(y_test == 0).sum():,}")
print(f"  Fraud:      {(y_test == 1).sum():,}")
```

## Section 5 — Apply SMOTE

# Purpose
The Synthetic Minority Oversampling Technique (SMOTE) is applied to address the severe class imbalance identified in Section 3. Without intervention, models trained on imbalanced data tend to be biased toward the majority class, resulting in poor fraud detection performance.

# How SMOTE Works
SMOTE generates synthetic fraud samples by interpolating between existing fraud cases in feature space, rather than simply duplicating them. This produces a more generalisable training set than simple oversampling.

# Critical Rule
> **SMOTE is applied to the training set only.**

Applying SMOTE to the test set would introduce synthetic data into the evaluation process, producing artificially optimistic results that do not reflect real-world performance. The test set must always represent the true class distribution.

# Expected Outcome
After SMOTE, the training set will contain an equal number of fraudulent and legitimate transactions, eliminating the class imbalance for the purpose of model training.

```python
# =============================================================================
# SECTION 5 — APPLY SMOTE
# =============================================================================

# Apply SMOTE to training data only
# The test set must remain untouched to ensure an unbiased evaluation
smote = SMOTE(random_state=RANDOM_SEED)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

print(f"Before SMOTE:")
print(f"  Training set shape:  {X_train.shape}")
print(f"  Legitimate: {(y_train == 0).sum():,}")
print(f"  Fraud:      {(y_train == 1).sum():,}")
print(f"\nAfter SMOTE:")
print(f"  Training set shape:  {X_train_resampled.shape}")
print(f"  Legitimate: {(y_train_resampled == 0).sum():,}")
print(f"  Fraud:      {(y_train_resampled == 1).sum():,}")
```

## Section 6 — Train Models

# Purpose
Three machine learning models are trained on the SMOTE-resampled training data. Each model represents a different approach to classification, allowing for a meaningful comparative evaluation.

# Models Selected
| Model | Rationale |
|---|---|
| Logistic Regression | Linear baseline — establishes a performance floor |
| Random Forest | Ensemble of decision trees — robust to imbalance and noise |
| Gradient Boosting | Sequential ensemble — strong performance on structured data |

# Hyperparameters
| Model | Parameter | Value | Reason |
|---|---|---|---|
| Logistic Regression | max_iter | 1000 | Ensures convergence on large dataset |
| Random Forest | n_estimators | 100 | Sufficient trees for stable predictions |
| Random Forest | n_jobs | -1 | Utilises all available CPU cores |
| Gradient Boosting | n_estimators | 100 | Balanced speed and performance |
| Gradient Boosting | max_depth | 3 | Prevents overfitting |

# Fixed Seed
All models use **random_state=42** to ensure reproducibility of results.

```python
# =============================================================================
# SECTION 6 — TRAIN MODELS
# =============================================================================

# NOTE: n_estimators set to 50 for development speed.
# Change to 100 for the final submission version.

import time

# --- Logistic Regression ---
print("Training Logistic Regression...")
start = time.time()
lr_model = LogisticRegression(random_state=RANDOM_SEED, max_iter=1000)
lr_model.fit(X_train_resampled, y_train_resampled)
lr_time = time.time() - start
print(f"  Completed in {lr_time:.1f} seconds")

# --- Random Forest ---
print("\nTraining Random Forest...")
start = time.time()
rf_model = RandomForestClassifier(n_estimators=50, random_state=RANDOM_SEED, n_jobs=-1)
rf_model.fit(X_train_resampled, y_train_resampled)
rf_time = time.time() - start
print(f"  Completed in {rf_time:.1f} seconds")

# --- Gradient Boosting ---
print("\nTraining Gradient Boosting...")
start = time.time()
gb_model = GradientBoostingClassifier(n_estimators=50, max_depth=3, random_state=RANDOM_SEED)
gb_model.fit(X_train_resampled, y_train_resampled)
gb_time = time.time() - start
print(f"  Completed in {gb_time:.1f} seconds")

print("\nAll models trained successfully.")
```

## Section 7 — Evaluate Models

# Purpose
Each trained model is evaluated against the held-out test set. The test set has never been seen by any model during training or resampling, ensuring an unbiased assessment of real-world performance.

# Evaluation Metrics
| Metric | Definition | Why It Matters |
|---|---|---|
| Precision | Of all fraud predictions, how many were correct | Measures false alarm rate |
| Recall | Of all actual frauds, how many were detected | Measures fraud detection rate |
| F1-Score | Harmonic mean of precision and recall | Primary balance metric |
| ROC-AUC | Area under the ROC curve | Overall discrimination ability |
| PR-AUC | Area under the Precision-Recall curve | **Primary metric** — most informative for imbalanced datasets |

# Why PR-AUC is the Primary Metric
ROC-AUC can be misleadingly optimistic on imbalanced datasets because it accounts for true negatives, which are abundant when fraud is rare. PR-AUC focuses exclusively on the minority class performance, making it a more honest measure of fraud detection capability.

```python
# =============================================================================
# SECTION 7 — EVALUATE MODELS
# =============================================================================

def evaluate_model(name, model, X_test, y_test):
    """
    Generates predictions and computes evaluation metrics for a given model.
    Returns a dictionary of metrics for later comparison.
    """
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    precision = classification_report(y_test, y_pred, output_dict=True)['1']['precision']
    recall    = classification_report(y_test, y_pred, output_dict=True)['1']['recall']
    f1        = classification_report(y_test, y_pred, output_dict=True)['1']['f1-score']
    roc_auc   = roc_auc_score(y_test, y_prob)
    pr_auc    = average_precision_score(y_test, y_prob)

    print(f"\n{'='*50}")
    print(f"Model: {name}")
    print(f"{'='*50}")
    print(classification_report(y_test, y_pred, target_names=['Legitimate', 'Fraud']))
    print(f"ROC-AUC : {roc_auc:.4f}")
    print(f"PR-AUC  : {pr_auc:.4f}")

    return {
        'Model'    : name,
        'Precision': round(precision, 4),
        'Recall'   : round(recall, 4),
        'F1'       : round(f1, 4),
        'ROC-AUC'  : round(roc_auc, 4),
        'PR-AUC'   : round(pr_auc, 4)
    }

# Evaluate all three models
lr_results = evaluate_model('Logistic Regression', lr_model, X_test, y_test)
rf_results = evaluate_model('Random Forest',       rf_model, X_test, y_test)
gb_results = evaluate_model('Gradient Boosting',   gb_model, X_test, y_test)
```

```python
# Display a clean summary comparison table of all three models
results_df = pd.DataFrame([lr_results, rf_results, gb_results])
results_df = results_df.set_index('Model')

print("Model Comparison Summary:")
print(results_df.to_string())
```

## Section 10 — Fraud Prediction Demo

# Purpose

# Demo Transactions
Two transactions are selected from the test set:
- **Transaction 1** — a known fraudulent transaction
- **Transaction 2** — a known legitimate transaction

The model returns a predicted label and a fraud probability score for each transaction, demonstrating both correct classification and the model's confidence level.

# Limitation
The dataset features V1–V28 are PCA-transformed and do not correspond to interpretable real-world attributes such as merchant name or location. As a result, individual transaction checking using human-readable inputs is not possible with this dataset. This limitation is documented as a critical reflection in the AT3 report, with IEEE-CIS identified as a future dataset that would support fully interpretable predictions.

```python
# =============================================================================
# SECTION 10 — FRAUD PREDICTION DEMO
# =============================================================================

# Extract one known fraudulent and one known legitimate transaction from the test set
fraud_indices      = y_test[y_test == 1].index
legitimate_indices = y_test[y_test == 0].index

fraud_sample      = X_test.loc[fraud_indices[0]].values.reshape(1, -1)
legitimate_sample = X_test.loc[legitimate_indices[0]].values.reshape(1, -1)

def predict_transaction(sample, actual_label):
    """
    Generates a fraud prediction and confidence score for a single transaction.
    """
    prediction   = rf_model.predict(sample)[0]
    probability  = rf_model.predict_proba(sample)[0][1]
    label        = 'FRAUD' if prediction == 1 else 'LEGITIMATE'
    actual       = 'FRAUD' if actual_label == 1 else 'LEGITIMATE'

    print(f"  Actual Label      : {actual}")
    print(f"  Predicted Label   : {label}")
    print(f"  Fraud Probability : {probability:.4f} ({probability * 100:.2f}%)")
    print(f"  Correct           : {'Yes' if prediction == actual_label else 'No'}")

print("=" * 45)
print("TRANSACTION 1 — Known Fraudulent")
print("=" * 45)
predict_transaction(fraud_sample, 1)

print("\n" + "=" * 45)
print("TRANSACTION 2 — Known Legitimate")
print("=" * 45)
predict_transaction(legitimate_sample, 0)
```

## Section 12 — Cross Validation

### Purpose
Cross validation is applied to verify that the evaluation results obtained in Section 7 are generalisable and not the product of a favourable train/test split.

### Methodology
- **Method:** 5-Fold Stratified Cross Validation
- **Stratification:** Each fold preserves the original fraud/legitimate class ratio
- **Metrics evaluated:** F1-Score, ROC-AUC, PR-AUC
- **Requirement satisfied:** FR-16 — Model Validation

### Why This Matters
A single train/test split can occasionally produce misleading results depending on how the data is divided. Cross validation trains and evaluates the model five times on different subsets of the data, and reports the mean and standard deviation of each metric. A low standard deviation confirms that the model performs consistently across different data splits.

```python
# =============================================================================
# SECTION 12 — CROSS VALIDATION
# =============================================================================

# 5-fold stratified cross validation is applied to all three models
# Stratified folds preserve the fraud/legitimate ratio in each fold
# This confirms that results from Section 7 are generalisable and not
# the result of a favourable train/test split

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_SEED)

cv_models = [
    ('Logistic Regression', lr_model),
    ('Random Forest',       rf_model),
    ('Gradient Boosting',   gb_model)
]

cv_results = []

print("Running 5-Fold Stratified Cross Validation...")
print("This may take several minutes.\n")

for name, model in cv_models:
    print(f"Evaluating: {name}...")

    f1_scores  = cross_val_score(model, X, y, cv=cv, scoring='f1',                n_jobs=-1)
    roc_scores = cross_val_score(model, X, y, cv=cv, scoring='roc_auc',           n_jobs=-1)
    pr_scores  = cross_val_score(model, X, y, cv=cv, scoring='average_precision', n_jobs=-1)

    cv_results.append({
        'Model'       : name,
        'F1 Mean'     : round(f1_scores.mean(),  4),
        'F1 Std'      : round(f1_scores.std(),   4),
        'ROC-AUC Mean': round(roc_scores.mean(), 4),
        'ROC-AUC Std' : round(roc_scores.std(),  4),
        'PR-AUC Mean' : round(pr_scores.mean(),  4),
        'PR-AUC Std'  : round(pr_scores.std(),   4)
    })

    print(f"  F1:      {f1_scores.mean():.4f} (+/- {f1_scores.std():.4f})")
    print(f"  ROC-AUC: {roc_scores.mean():.4f} (+/- {roc_scores.std():.4f})")
    print(f"  PR-AUC:  {pr_scores.mean():.4f} (+/- {pr_scores.std():.4f})\n")

# Display consolidated cross validation results table
cv_df = pd.DataFrame(cv_results).set_index('Model')
print("=" * 65)
print(f"{'5-FOLD CROSS VALIDATION RESULTS':^65}")
print("=" * 65)
print(cv_df.to_string())
print("=" * 65)
print("\nSection 12 complete.")
```

```python
# =============================================================================
# SECTION 14 — HYPERPARAMETER TUNING
# =============================================================================

# GridSearchCV with 3-fold cross validation is applied to Random Forest
# and Gradient Boosting using the SMOTE-resampled training data

cv_grid = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_SEED)

# --- Random Forest Parameter Grid ---
rf_param_grid = {
    'n_estimators' : [50, 100],
    'max_depth'    : [None, 10, 20],
    'min_samples_split': [2, 5]
}

# --- Gradient Boosting Parameter Grid ---
gb_param_grid = {
    'n_estimators' : [50, 100],
    'max_depth'    : [3, 5],
    'learning_rate': [0.05, 0.1]
}

print("Running GridSearchCV — Random Forest...")
print("This will take several minutes.\n")

rf_grid_search = GridSearchCV(
    RandomForestClassifier(random_state=RANDOM_SEED, n_jobs=-1),
    rf_param_grid,
    cv=cv_grid,
    scoring='f1',
    n_jobs=-1,
    verbose=1
)
rf_grid_search.fit(X_train_resampled, y_train_resampled)

print(f"\nBest parameters — Random Forest:")
print(f"  {rf_grid_search.best_params_}")
print(f"Best CV F1 Score: {rf_grid_search.best_score_:.4f}")

print("\nRunning GridSearchCV — Gradient Boosting...")
print("This will take several minutes.\n")

gb_grid_search = GridSearchCV(
    GradientBoostingClassifier(random_state=RANDOM_SEED),
    gb_param_grid,
    cv=cv_grid,
    scoring='f1',
    n_jobs=-1,
    verbose=1
)
gb_grid_search.fit(X_train_resampled, y_train_resampled)

print(f"\nBest parameters — Gradient Boosting:")
print(f"  {gb_grid_search.best_params_}")
print(f"Best CV F1 Score: {gb_grid_search.best_score_:.4f}")
```

## Section 16 — External Dataset Prediction

# Purpose
This section demonstrates the trained Random Forest model operating as a deployable prediction engine. Two input methods are supported — CSV file upload for batch prediction and manual entry for single transaction prediction. This simulates how the model would function in a real-world fraud detection system.

# Input Requirements
For CSV upload, the file must contain the following columns in any order:
`Time`, `V1` through `V28`, `Amount`

A `Class` column is not required — the model generates predictions independently.

# Output
- Predicted label for each transaction (FRAUD or LEGITIMATE)
- Fraud probability score (0.0 to 1.0)
- Summary statistics across all predictions
- Results exported as a CSV file to the notebook directory

```python
# =============================================================================
# SECTION 16 — EXTERNAL DATASET PREDICTION
# =============================================================================

import os

# Load the saved model and scaler from disk
loaded_model  = joblib.load('../model/random_forest_model.pkl')
loaded_scaler = joblib.load('../model/scaler.pkl')

# Define the required columns for any input dataset
REQUIRED_COLUMNS = ['Time', 'V1', 'V2', 'V3', 'V4', 'V5', 'V6', 'V7', 'V8',
                    'V9', 'V10', 'V11', 'V12', 'V13', 'V14', 'V15', 'V16',
                    'V17', 'V18', 'V19', 'V20', 'V21', 'V22', 'V23', 'V24',
                    'V25', 'V26', 'V27', 'V28', 'Amount']

def preprocess_input(df_input):
    """
    Validates and preprocesses an input DataFrame for prediction.
    Applies the fitted StandardScaler to Amount and Time columns.
    """
    # Check all required columns are present
    missing = [col for col in REQUIRED_COLUMNS if col not in df_input.columns]
    if missing:
        raise ValueError(f"Missing required columns: {missing}")

    df_processed = df_input[REQUIRED_COLUMNS].copy()
    df_processed[['Amount', 'Time']] = loaded_scaler.transform(
        df_processed[['Amount', 'Time']]
    )
    return df_processed


def predict_csv(filepath):
    """
    Loads a CSV file, runs predictions on every row, and exports results.
    """
    if not os.path.exists(filepath):
        print(f"File not found: {filepath}")
        return

    print(f"Loading file: {filepath}")
    df_input = pd.read_csv(filepath)
    print(f"Rows loaded: {len(df_input):,}")

    # Preprocess and predict
    df_processed  = preprocess_input(df_input)
    predictions   = loaded_model.predict(df_processed)
    probabilities = loaded_model.predict_proba(df_processed)[:, 1]

    # Attach results to original dataframe
    df_input['Predicted_Label']       = predictions
    df_input['Fraud_Probability']     = probabilities.round(4)
    df_input['Predicted_Class_Label'] = df_input['Predicted_Label'].map(
        {0: 'LEGITIMATE', 1: 'FRAUD'}
    )

    # Summary statistics
    total      = len(df_input)
    fraud      = (predictions == 1).sum()
    legitimate = (predictions == 0).sum()
    fraud_rate = (fraud / total) * 100

    print(f"\n{'='*50}")
    print(f"  PREDICTION SUMMARY")
    print(f"{'='*50}")
    print(f"  Total Transactions : {total:,}")
    print(f"  Legitimate         : {legitimate:,}")
    print(f"  Fraud Detected     : {fraud:,}")
    print(f"  Fraud Rate         : {fraud_rate:.2f}%")
    print(f"{'='*50}")

    # Show top 10 highest risk transactions
    print(f"\nTop 10 Highest Risk Transactions:")
    top_fraud = df_input.nlargest(10, 'Fraud_Probability')[
        ['Amount', 'Fraud_Probability', 'Predicted_Class_Label']
    ]
    print(top_fraud.to_string(index=False))

    # Export results to CSV
    output_path = '../notebook/prediction_results.csv'
    df_input.to_csv(output_path, index=False)
    print(f"\nResults exported → {output_path}")

    return df_input


def predict_single(time, amount, v_features):
    """
    Predicts fraud probability for a single manually entered transaction.
    v_features must be a list of 28 values corresponding to V1 through V28.
    """
    if len(v_features) != 28:
        print("Error: exactly 28 V feature values are required.")
        return

    # Build a single row DataFrame
    columns    = REQUIRED_COLUMNS
    values     = [time] + v_features + [amount]
    df_single  = pd.DataFrame([values], columns=columns)

    df_processed  = preprocess_input(df_single)
    prediction    = loaded_model.predict(df_processed)[0]
    probability   = loaded_model.predict_proba(df_processed)[0][1]
    label         = 'FRAUD' if prediction == 1 else 'LEGITIMATE'

    print(f"\n{'='*50}")
    print(f"  SINGLE TRANSACTION PREDICTION")
    print(f"{'='*50}")
    print(f"  Amount             : {amount}")
    print(f"  Time               : {time}")
    print(f"  Predicted Label    : {label}")
    print(f"  Fraud Probability  : {probability:.4f} ({probability*100:.2f}%)")
    risk = 'HIGH' if probability >= 0.7 else 'MEDIUM' if probability >= 0.3 else 'LOW'
    print(f"  Risk Level         : {risk}")
    print(f"{'='*50}")


print("Prediction functions loaded successfully.")
print("\nUsage:")
print("  CSV    → predict_csv('path/to/your/file.csv')")
print("  Single → predict_single(time=0, amount=150.00, v_features=[v1,v2,...,v28])")
```

## Section 17 — Multi-Domain Fraud Detection: Three Specialist Models

This section extends the fraud detection pipeline beyond the ULB dataset by training
three domain-specialist models, each optimised for a specific type of financial fraud.
The system automatically routes transactions to the most appropriate model based on
transaction type — eliminating the need for users to select a model manually.

**Three Specialist Models:**

| Model | Dataset | Fraud Domain | Transactions |
|---|---|---|---|
| Credit Card Fraud | Sparkov (Kaggle) | In-person & online card purchases | 1,852,394 |
| Online Payment Fraud | IEEE-CIS (Kaggle) | E-commerce & digital transactions | 590,540 |
| Mobile Transfer Fraud | PaySim (Kaggle) | Mobile money & bank transfers | 6,362,620 |

**Auto-Routing Logic:**
The system determines the appropriate specialist model from the transaction type field:
- Card Purchase / ATM / Retail → Credit Card model (Sparkov)
- Online Payment / E-commerce → Online Payment model (IEEE-CIS)
- Bank Transfer / Mobile Payment → Mobile Transfer model (PaySim)

**Universal Features (shared across all models):**
- `amount`, `amount_log`, `is_high_amount`, `is_micro` — transaction value signals
- `hour`, `is_night`, `is_weekend` — temporal fraud patterns
- `merchant_category` — encoded transaction category
- `is_international`, `is_new_merchant` — transaction context flags
- `balance_change` — transaction impact on account balance

**Additional domain-specific features per model:**
- Sparkov: `geo_distance`, `is_far_merchant`, `gender`, `age`, `city_pop_log`, `is_online`
- IEEE-CIS: `card_type`, `card_category`, `suspicious_email`, `count_c1–c14`, `id_01–id_20`, `d1–d15`
- PaySim: `balance_change` (primary signal alongside amount and transaction type)

**Methodology (consistent across all three models):**
- Stratified 80/20 train-test split (random seed 42)
- Combined sampling: RandomUnderSampler (5% fraud ratio) followed by SMOTE (1:1 balance)
- Random Forest Classifier (200–500 estimators)
- Optimal threshold selection via Precision-Recall curve maximising F1 score

Unlike the anonymised V1-V28 PCA features in Sections 1-16, all features used here
are fully interpretable and available from any real-world payment processing system.

```python
#Imports and Setup
import gc
import json
import warnings
warnings.filterwarnings('ignore')
from sklearn.metrics import precision_recall_curve
from imblearn.under_sampling import RandomUnderSampler

gc.collect()
print("Section 17 — Multi-Domain Fraud Detection\n")

RANDOM_SEED = 42

def evaluate_model(name, y_true, y_pred, y_proba):
    from sklearn.metrics import (precision_score, recall_score, f1_score,
                                  roc_auc_score, average_precision_score,
                                  confusion_matrix)
    p   = precision_score(y_true, y_pred, zero_division=0)
    r   = recall_score(y_true, y_pred, zero_division=0)
    f1  = f1_score(y_true, y_pred, zero_division=0)
    roc = roc_auc_score(y_true, y_proba)
    pr  = average_precision_score(y_true, y_proba)
    cm  = confusion_matrix(y_true, y_pred)
    fp  = cm[0][1]
    print(f"\n{'='*50}")
    print(f"  {name}")
    print(f"{'='*50}")
    print(f"  Precision      : {p:.4f}")
    print(f"  Recall         : {r:.4f}")
    print(f"  F1             : {f1:.4f}")
    print(f"  ROC-AUC        : {roc:.4f}")
    print(f"  PR-AUC         : {pr:.4f}")
    print(f"  False Positives: {fp}")
    return {'precision': p, 'recall': r, 'f1': f1, 'roc_auc': roc, 'pr_auc': pr}

def get_optimal_threshold(y_true, y_proba):
    p, r, t = precision_recall_curve(y_true, y_proba)
    f1s     = 2 * (p * r) / (p + r + 1e-8)
    idx     = f1s.argmax()
    return t[idx]

print("Helper functions ready.")
```

```python
#Model 1: Credit Card Fraud (Sparkov)
print("="*60)
print("MODEL 1 — Credit Card Fraud (Sparkov)")
print("="*60)

sparkov_train = pd.read_csv('../data/sparkov_train.csv')
sparkov_test  = pd.read_csv('../data/sparkov_test.csv')
sparkov_all   = pd.concat([sparkov_train, sparkov_test], ignore_index=True)
del sparkov_train, sparkov_test
gc.collect()

def extract_sparkov_features(df):
    out = pd.DataFrame()
    out['amount']            = df['amt']
    out['amount_log']        = np.log1p(df['amt'])
    out['is_high_amount']    = (df['amt'] > 1000).astype(int)
    out['is_micro']          = (df['amt'] < 1).astype(int)
    dt                       = pd.to_datetime(df['trans_date_trans_time'])
    out['hour']              = dt.dt.hour
    out['is_night']          = out['hour'].between(0, 5).astype(int)
    out['is_weekend']        = dt.dt.dayofweek.isin([5, 6]).astype(int)
    cat_map = {
        'grocery_pos': 0, 'grocery_net': 0,
        'shopping_pos': 1, 'shopping_net': 1,
        'misc_pos': 2, 'misc_net': 2,
        'food_dining': 3, 'entertainment': 3,
        'gas_transport': 4, 'travel': 4,
        'health_fitness': 5, 'personal_care': 5,
        'home': 6, 'kids_pets': 6,
        'cash_advance': 7
    }
    out['merchant_category'] = df['category'].map(cat_map).fillna(2).astype(int)
    out['gender']            = (df['gender'] == 'M').astype(int)
    dob                      = pd.to_datetime(df['dob'], errors='coerce')
    out['age']               = ((dt - dob).dt.days / 365).fillna(35).astype(int)
    out['city_pop_log']      = np.log1p(df['city_pop'])
    out['geo_distance']      = np.sqrt(
        (df['lat'] - df['merch_lat'])**2 +
        (df['long'] - df['merch_long'])**2)
    out['is_far_merchant']   = (out['geo_distance'] > 2).astype(int)
    out['is_international']  = (out['geo_distance'] > 5).astype(int)
    out['is_new_merchant']   = (df['category'] == 'misc_net').astype(int)
    out['balance_change']    = df['amt']
    out['is_online']         = df['category'].str.contains('net', na=False).astype(int)
    out['label']             = df['is_fraud']
    return out

SPARKOV_FEATURES = [
    'amount', 'amount_log', 'is_high_amount', 'is_micro',
    'hour', 'is_night', 'is_weekend', 'merchant_category',
    'is_international', 'is_new_merchant', 'balance_change',
    'geo_distance', 'is_far_merchant', 'is_online',
    'gender', 'age', 'city_pop_log'
]
SPARKOV_SCALE = ['amount', 'balance_change', 'geo_distance', 'age', 'city_pop_log']

df_sp = extract_sparkov_features(sparkov_all).dropna()
del sparkov_all
gc.collect()
print(f"Sparkov: {len(df_sp):,} rows | Fraud: {df_sp['label'].sum():,}")

X_sp = df_sp[SPARKOV_FEATURES]
y_sp = df_sp['label']
del df_sp
gc.collect()

X_sp_tr, X_sp_te, y_sp_tr, y_sp_te = train_test_split(
    X_sp, y_sp, test_size=0.2, random_state=RANDOM_SEED, stratify=y_sp)

sc_sp      = StandardScaler()
X_sp_tr_sc = X_sp_tr.copy()
X_sp_te_sc = X_sp_te.copy()
X_sp_tr_sc[SPARKOV_SCALE] = sc_sp.fit_transform(X_sp_tr[SPARKOV_SCALE])
X_sp_te_sc[SPARKOV_SCALE] = sc_sp.transform(X_sp_te[SPARKOV_SCALE])
del X_sp_tr, X_sp_te
gc.collect()

rus1             = RandomUnderSampler(sampling_strategy=0.05, random_state=RANDOM_SEED)
X_sp_un, y_sp_un = rus1.fit_resample(X_sp_tr_sc, y_sp_tr)
sm1              = SMOTE(random_state=RANDOM_SEED)
X_sp_bal, y_sp_bal = sm1.fit_resample(X_sp_un, y_sp_un)
del X_sp_un, y_sp_un
gc.collect()
print(f"After sampling: {X_sp_bal.shape}")

print("Training Random Forest...")
rf_sp = RandomForestClassifier(
    n_estimators=200, max_depth=None,
    min_samples_leaf=1, random_state=RANDOM_SEED, n_jobs=-1)
rf_sp.fit(X_sp_bal, y_sp_bal)
del X_sp_bal, y_sp_bal
gc.collect()

rf_sp_proba = rf_sp.predict_proba(X_sp_te_sc)[:, 1]
rf_sp_thr   = get_optimal_threshold(y_sp_te, rf_sp_proba)
rf_sp_pred  = (rf_sp_proba >= rf_sp_thr).astype(int)
print(f"Optimal threshold: {rf_sp_thr:.4f}")
sp_metrics  = evaluate_model('Credit Card RF (Sparkov)', y_sp_te, rf_sp_pred, rf_sp_proba)
del X_sp_te_sc
gc.collect()
```

```python
#Model 2: Online Payment Fraud (IEEE-CIS)
print("="*60)
print("MODEL 2 — Online Payment Fraud (IEEE-CIS)")
print("="*60)

ieee = pd.read_csv('../data/ieee_cis.csv')
print(f"Loaded: {ieee.shape}")

def extract_ieee_features_v3(df):
    out = pd.DataFrame()
    out['amount']            = df['TransactionAmt']
    out['amount_log']        = np.log1p(df['TransactionAmt'])
    out['is_high_amount']    = (df['TransactionAmt'] > 1000).astype(int)
    out['is_micro']          = (df['TransactionAmt'] < 1).astype(int)
    out['hour']              = (df['TransactionDT'] / 3600 % 24).astype(int)
    out['is_night']          = out['hour'].between(0, 5).astype(int)
    out['is_weekend']        = ((df['TransactionDT'] / 86400) % 7 >= 5).astype(int)
    prod_map                 = {'W': 0, 'H': 1, 'C': 2, 'S': 3, 'R': 4}
    out['merchant_category'] = df['ProductCD'].map(prod_map).fillna(0).astype(int)
    card4_map                = {'visa': 0, 'mastercard': 1,
                                'american express': 2, 'discover': 3}
    out['card_type']         = df['card4'].map(card4_map).fillna(-1).astype(int)
    card6_map                = {'debit': 0, 'credit': 1,
                                'debit or credit': 2, 'charge card': 3}
    out['card_category']     = df['card6'].map(card6_map).fillna(-1).astype(int)
    out['card1']             = df['card1'].fillna(0)
    out['card2']             = df['card2'].fillna(0)
    out['is_international']  = (df['addr2'] != 87).astype(int)
    out['geo_distance']      = df['dist1'].fillna(0)
    out['is_far_merchant']   = (df['dist1'].fillna(0) > 100).astype(int)
    suspicious               = ['anonymous.com','gmail.com','yahoo.com','hotmail.com']
    out['suspicious_email']  = df['P_emaildomain'].isin(suspicious).astype(int)
    out['email_match']       = (df['P_emaildomain'] == df['R_emaildomain']).astype(int)
    out['count_match']       = df['C1'].fillna(0)
    out['count_distinct']    = df['C2'].fillna(0)
    out['count_c3']          = df['C3'].fillna(0)
    out['count_c4']          = df['C4'].fillna(0)
    out['count_c5']          = df['C5'].fillna(0)
    out['count_c6']          = df['C6'].fillna(0)
    out['count_c9']          = df['C9'].fillna(0)
    out['count_c11']         = df['C11'].fillna(0)
    out['count_c13']         = df['C13'].fillna(0)
    out['count_c14']         = df['C14'].fillna(0)
    out['id_01']             = df['id_01'].fillna(0) if 'id_01' in df.columns else 0
    out['id_02']             = df['id_02'].fillna(0) if 'id_02' in df.columns else 0
    out['id_03']             = df['id_03'].fillna(0) if 'id_03' in df.columns else 0
    out['id_05']             = df['id_05'].fillna(0) if 'id_05' in df.columns else 0
    out['id_06']             = df['id_06'].fillna(0) if 'id_06' in df.columns else 0
    out['id_09']             = df['id_09'].fillna(0) if 'id_09' in df.columns else 0
    out['id_10']             = df['id_10'].fillna(0) if 'id_10' in df.columns else 0
    out['id_11']             = df['id_11'].fillna(0) if 'id_11' in df.columns else 0
    out['id_13']             = df['id_13'].fillna(0) if 'id_13' in df.columns else 0
    out['id_17']             = df['id_17'].fillna(0) if 'id_17' in df.columns else 0
    out['id_19']             = df['id_19'].fillna(0) if 'id_19' in df.columns else 0
    out['id_20']             = df['id_20'].fillna(0) if 'id_20' in df.columns else 0
    out['d1']                = df['D1'].fillna(0) if 'D1' in df.columns else 0
    out['d2']                = df['D2'].fillna(0) if 'D2' in df.columns else 0
    out['d3']                = df['D3'].fillna(0) if 'D3' in df.columns else 0
    out['d4']                = df['D4'].fillna(0) if 'D4' in df.columns else 0
    out['d10']               = df['D10'].fillna(0) if 'D10' in df.columns else 0
    out['d15']               = df['D15'].fillna(0) if 'D15' in df.columns else 0
    out['is_new_merchant']   = (df['ProductCD'] == 'C').astype(int)
    out['balance_change']    = df['TransactionAmt']
    out['is_online']         = df['ProductCD'].isin(['C','H','R']).astype(int)
    out['label']             = df['isFraud']
    return out

IEEE_FEATURES_V3 = [
    'amount', 'amount_log', 'is_high_amount', 'is_micro',
    'hour', 'is_night', 'is_weekend', 'merchant_category',
    'is_international', 'is_new_merchant', 'balance_change',
    'card_type', 'card_category', 'card1', 'card2',
    'geo_distance', 'is_far_merchant',
    'suspicious_email', 'email_match', 'is_online',
    'count_match', 'count_distinct', 'count_c3', 'count_c4',
    'count_c5', 'count_c6', 'count_c9', 'count_c11',
    'count_c13', 'count_c14',
    'id_01', 'id_02', 'id_03', 'id_05', 'id_06',
    'id_09', 'id_10', 'id_11', 'id_13', 'id_17',
    'id_19', 'id_20',
    'd1', 'd2', 'd3', 'd4', 'd10', 'd15'
]
IEEE_SCALE_V3 = [
    'amount', 'balance_change', 'geo_distance',
    'count_match', 'count_distinct', 'count_c3', 'count_c4',
    'count_c5', 'count_c6', 'count_c9', 'count_c11',
    'count_c13', 'count_c14', 'card1', 'card2',
    'id_01', 'id_02', 'id_03', 'id_05', 'id_06',
    'id_09', 'id_10', 'id_11', 'id_13', 'id_17',
    'id_19', 'id_20', 'd1', 'd2', 'd3', 'd4', 'd10', 'd15'
]

df_ie_v3 = extract_ieee_features_v3(ieee).dropna()
del ieee
gc.collect()
print(f"IEEE-CIS: {len(df_ie_v3):,} rows | Fraud: {df_ie_v3['label'].sum():,}")

X_ie3 = df_ie_v3[IEEE_FEATURES_V3]
y_ie3 = df_ie_v3['label']
del df_ie_v3
gc.collect()

X_ie3_tr, X_ie3_te, y_ie3_tr, y_ie3_te = train_test_split(
    X_ie3, y_ie3, test_size=0.2, random_state=RANDOM_SEED, stratify=y_ie3)

sc_ie3       = StandardScaler()
X_ie3_tr_sc  = X_ie3_tr.copy()
X_ie3_te_sc  = X_ie3_te.copy()
X_ie3_tr_sc[IEEE_SCALE_V3] = sc_ie3.fit_transform(X_ie3_tr[IEEE_SCALE_V3])
X_ie3_te_sc[IEEE_SCALE_V3] = sc_ie3.transform(X_ie3_te[IEEE_SCALE_V3])
del X_ie3_tr, X_ie3_te
gc.collect()

rus2               = RandomUnderSampler(sampling_strategy=0.05, random_state=RANDOM_SEED)
X_ie3_un, y_ie3_un = rus2.fit_resample(X_ie3_tr_sc, y_ie3_tr)
sm2                = SMOTE(random_state=RANDOM_SEED)
X_ie3_bal, y_ie3_bal = sm2.fit_resample(X_ie3_un, y_ie3_un)
del X_ie3_un, y_ie3_un
gc.collect()
print(f"After sampling: {X_ie3_bal.shape}")

print("Training Random Forest (initial pass for feature selection)...")
rf_ie_init = RandomForestClassifier(
    n_estimators=100, max_depth=None,
    min_samples_leaf=1, class_weight='balanced_subsample',
    random_state=RANDOM_SEED, n_jobs=-1)
rf_ie_init.fit(X_ie3_bal, y_ie3_bal)

# Select top 30 features
importances      = pd.Series(rf_ie_init.feature_importances_, index=IEEE_FEATURES_V3)
top_features     = importances.sort_values(ascending=False).head(30).index.tolist()
IEEE_SCALE_FINAL = [f for f in IEEE_SCALE_V3 if f in top_features]
print(f"\nTop 30 features selected.")

# Retrain on top features only
X_ie3_bal_top = X_ie3_bal[top_features] if hasattr(X_ie3_bal, '__getitem__') else pd.DataFrame(X_ie3_bal, columns=IEEE_FEATURES_V3)[top_features]
X_ie3_te_top  = X_ie3_te_sc[top_features]

del X_ie3_bal
gc.collect()

print("Training final Random Forest on top 30 features...")
rf_ie5 = RandomForestClassifier(
    n_estimators=500, max_depth=None,
    min_samples_leaf=1, class_weight='balanced_subsample',
    random_state=RANDOM_SEED, n_jobs=-1)
rf_ie5.fit(X_ie3_bal_top, y_ie3_bal)
del X_ie3_bal_top, y_ie3_bal
gc.collect()

rf_ie5_proba = rf_ie5.predict_proba(X_ie3_te_top)[:, 1]
rf_ie5_thr   = get_optimal_threshold(y_ie3_te, rf_ie5_proba)
rf_ie5_pred  = (rf_ie5_proba >= rf_ie5_thr).astype(int)
print(f"Optimal threshold: {rf_ie5_thr:.4f}")
ie_metrics   = evaluate_model('Online Payment RF (IEEE-CIS)', y_ie3_te, rf_ie5_pred, rf_ie5_proba)
del X_ie3_te_sc, X_ie3_te_top
gc.collect()
```

```python
# Model 3: Mobile Transfer Fraud (PaySim)
print("="*60)
print("MODEL 3 — Mobile Transfer Fraud (PaySim)")
print("="*60)

paysim       = pd.read_csv('../data/paysim.csv',
                   usecols=['step','type','amount','oldbalanceOrg',
                            'newbalanceOrig','isFraud'])
paysim_fraud = paysim[paysim['isFraud'] == 1]
paysim_legit = paysim[paysim['isFraud'] == 0].sample(n=50000, random_state=RANDOM_SEED)
ps_sample    = pd.concat([paysim_fraud, paysim_legit])
del paysim, paysim_fraud, paysim_legit
gc.collect()

def extract_paysim_features(df):
    out = pd.DataFrame()
    out['amount']            = df['amount']
    out['amount_log']        = np.log1p(df['amount'])
    out['is_high_amount']    = (df['amount'] > 1000).astype(int)
    out['is_micro']          = (df['amount'] < 1).astype(int)
    out['hour']              = (df['step'] % 24).astype(int)
    out['is_night']          = out['hour'].between(0, 5).astype(int)
    out['is_weekend']        = ((df['step'] / 24) % 7 >= 5).astype(int)
    type_map                 = {'CASH_OUT': 0, 'PAYMENT': 1, 'CASH_IN': 2,
                                'TRANSFER': 3, 'DEBIT': 4}
    out['merchant_category'] = df['type'].map(type_map).fillna(2).astype(int)
    out['is_international']  = (df['type'] == 'TRANSFER').astype(int)
    out['is_new_merchant']   = (df['type'] == 'CASH_OUT').astype(int)
    out['balance_change']    = (df['newbalanceOrig'] - df['oldbalanceOrg']).abs()
    out['label']             = df['isFraud']
    return out

PAYSIM_FEATURES = [
    'amount', 'amount_log', 'is_high_amount', 'is_micro',
    'hour', 'is_night', 'is_weekend', 'merchant_category',
    'is_international', 'is_new_merchant', 'balance_change'
]
PAYSIM_SCALE = ['amount', 'balance_change']

df_ps = extract_paysim_features(ps_sample).dropna()
del ps_sample
gc.collect()
print(f"PaySim: {len(df_ps):,} rows | Fraud: {df_ps['label'].sum():,}")

X_ps = df_ps[PAYSIM_FEATURES]
y_ps = df_ps['label']
del df_ps
gc.collect()

X_ps_tr, X_ps_te, y_ps_tr, y_ps_te = train_test_split(
    X_ps, y_ps, test_size=0.2, random_state=RANDOM_SEED, stratify=y_ps)

sc_ps      = StandardScaler()
X_ps_tr_sc = X_ps_tr.copy()
X_ps_te_sc = X_ps_te.copy()
X_ps_tr_sc[PAYSIM_SCALE] = sc_ps.fit_transform(X_ps_tr[PAYSIM_SCALE])
X_ps_te_sc[PAYSIM_SCALE] = sc_ps.transform(X_ps_te[PAYSIM_SCALE])
del X_ps_tr, X_ps_te
gc.collect()

print(f"Fraud rate: {y_ps_tr.mean()*100:.1f}% — applying SMOTE only...")
sm3                = SMOTE(random_state=RANDOM_SEED)
X_ps_bal, y_ps_bal = sm3.fit_resample(X_ps_tr_sc, y_ps_tr)
gc.collect()
print(f"After SMOTE: {X_ps_bal.shape}")

print("Training Random Forest...")
rf_ps = RandomForestClassifier(
    n_estimators=200, max_depth=None,
    min_samples_leaf=1, random_state=RANDOM_SEED, n_jobs=-1)
rf_ps.fit(X_ps_bal, y_ps_bal)
del X_ps_bal, y_ps_bal
gc.collect()

rf_ps_proba = rf_ps.predict_proba(X_ps_te_sc)[:, 1]
rf_ps_thr   = get_optimal_threshold(y_ps_te, rf_ps_proba)
rf_ps_pred  = (rf_ps_proba >= rf_ps_thr).astype(int)
print(f"Optimal threshold: {rf_ps_thr:.4f}")
ps_metrics  = evaluate_model('Mobile Transfer RF (PaySim)', y_ps_te, rf_ps_pred, rf_ps_proba)
del X_ps_te_sc
gc.collect()
```

```python
#Save All Models
print("Saving all models...\n")

joblib.dump(rf_sp,  '../model/model_credit_card.pkl',     compress=3)
joblib.dump(sc_sp,  '../model/scaler_credit_card.pkl',    compress=3)
joblib.dump(rf_ie5, '../model/model_online_payment.pkl',  compress=3)
joblib.dump(sc_ie3, '../model/scaler_online_payment.pkl', compress=3)
joblib.dump(rf_ps,  '../model/model_mobile_transfer.pkl', compress=3)
joblib.dump(sc_ps,  '../model/scaler_mobile_transfer.pkl',compress=3)

model_config = {
    'credit_card': {
        'features'         : SPARKOV_FEATURES,
        'scale_cols'       : SPARKOV_SCALE,
        'threshold'        : float(rf_sp_thr),
        'dataset'          : 'Sparkov',
        'description'      : 'Credit card purchases — in-person and online retail',
        'transaction_types': ['card_purchase', 'atm', 'online_shopping',
                              'contactless', 'grocery', 'entertainment']
    },
    'online_payment': {
        'features'         : top_features,
        'scale_cols'       : IEEE_SCALE_FINAL,
        'threshold'        : float(rf_ie5_thr),
        'dataset'          : 'IEEE-CIS',
        'description'      : 'Online payments — e-commerce and digital transactions',
        'transaction_types': ['ecommerce', 'digital_payment', 'subscription',
                              'online_transfer', 'card_not_present']
    },
    'mobile_transfer': {
        'features'         : PAYSIM_FEATURES,
        'scale_cols'       : PAYSIM_SCALE,
        'threshold'        : float(rf_ps_thr),
        'dataset'          : 'PaySim',
        'description'      : 'Mobile money transfers and bank payments',
        'transaction_types': ['bank_transfer', 'mobile_payment',
                              'cash_withdrawal', 'direct_debit']
    }
}

with open('../model/model_config.json', 'w') as f:
    json.dump(model_config, f, indent=2)

print("Model sizes:")
for name in ['model_credit_card', 'model_online_payment', 'model_mobile_transfer']:
    size = os.path.getsize(f'../model/{name}.pkl') / 1024 / 1024
    print(f"  {name}.pkl — {size:.2f} MB")

print("\nmodel_config.json saved.")
```
